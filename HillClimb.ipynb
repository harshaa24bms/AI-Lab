```python
def get_cost(state):
    cost = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1
    return cost

def print_board(state):
    n = len(state)
    for r in range(n):
        row_str = []
        for c in range(n):
            if state[c] == r:
                row_str.append("Q")
            else:
                row_str.append(".")
        print(" ".join(row_str))

def hill_climbing():
    current_state = [3, 1, 2, 0]
    n = len(current_state)

    while True:
        current_cost = get_cost(current_state)
        state_str = ", ".join(f"x_{i}={current_state[i]}" for i in range(n))
        print(f"Neighbours of {state_str} (cost is {current_cost})\n")

        best_neighbor = None
        best_cost = current_cost

        for i in range(n):
            for j in range(i + 1, n):
                neighbor = list(current_state)
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
                neighbor_cost = get_cost(neighbor)

                neighbor_str = ", ".join(f"x_{k}={neighbor[k]}" for k in range(n))

                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_neighbor = neighbor
                    print(f" • {neighbor_str} cost is {neighbor_cost}. Choose this neighbour.")
                else:
                    print(f" • {neighbor_str} cost is {neighbor_cost}.")

        print()
        if best_neighbor is None or best_cost >= current_cost:
            break

        current_state = best_neighbor
        if best_cost == 0:
            print("Goal reached! Final Board Matrix:")
            print_board(current_state)
            break

if __name__ == "__main__":
    hill_climbing()
```
