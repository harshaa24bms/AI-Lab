```python
# Define the dimensions of the puzzle
N = 3

class PuzzleState:
    def __init__(self, board, x, y, depth, parent=None):
        self.board = board
        self.x = x
        self.y = y
        self.depth = depth
        self.parent = parent

row = [0, 0, -1, 1]
col = [-1, 1, 0, 0]

def is_goal_state(board):
    goal = [[1, 2, 3],[8,0,4],[7,6,5]]
    return board == goal

def is_valid(x, y):
    return 0 <= x < N and 0 <= y < N

def print_board(board):
    for r in board:
        print(' '.join(map(str, r)))
    print("--------")

def depth_limited_search(start, x, y, limit):
    stack = []
    initial_state = PuzzleState(start, x, y, 0, None)
    stack.append((initial_state, {tuple(map(tuple, start))}))

    while stack:
        curr, branch_visited = stack.pop()

        if is_goal_state(curr.board):
            return curr  # Return the goal node to reconstruct the path

        if curr.depth >= limit:
            continue

        for i in range(4):
            new_x = curr.x + row[i]
            new_y = curr.y + col[i]

            if is_valid(new_x, new_y):
                new_board = [r[:] for r in curr.board]
                # Swap the tiles
                new_board[curr.x][curr.y], new_board[new_x][new_y] = new_board[new_x][new_y], new_board[curr.x][curr.y]

                board_tuple = tuple(map(tuple, new_board))

                if board_tuple not in branch_visited:
                    new_state = PuzzleState(new_board, new_x, new_y, curr.depth + 1, curr)
                    new_branch_visited = branch_visited.copy()
                    new_branch_visited.add(board_tuple)
                    stack.append((new_state, new_branch_visited))

    return None

def solve_puzzle_iddfs(start, x, y, max_depth=50):
    print('Initial State:')
    print_board(start)

    for depth_limit in range(max_depth + 1):
        print(f"Searching with Depth Limit: {depth_limit}...")
        goal_node = depth_limited_search(start, x, y, depth_limit)
        if goal_node:
            print(f'Goal state reached at depth {goal_node.depth}!')

            path = []
            curr = goal_node
            while curr:
                path.append(curr.board)
                curr = curr.parent

            path.reverse()
            print("\n--- Solution Path ---")
            for step, board in enumerate(path):
                print(f"Step {step}:")
                print_board(board)
            return

    print(f'No solution found within depth limit {max_depth}.')

# Driver Code
if __name__ == '__main__':

    start_state = [[2, 8, 3],[1,6,4],[7,0,5]]

    blank_x, blank_y = 2, 1

    solve_puzzle_iddfs(start_state, blank_x, blank_y, max_depth=10)
```
